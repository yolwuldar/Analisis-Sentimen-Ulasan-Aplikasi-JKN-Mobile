```python
from google_play_scraper import reviews, Sort   # import lbrary
import pandas as pd
```

# 2. Proses Scraping ulasan aplikasi Mobile JKN

```python
scrapreview, continuation_token = reviews(
    'app.bpjs.mobile',        # id dari aplikasi Mobile JKN
    lang='id',                # bahasa ulasan yang diambil
    country='id',             # negara
    sort=Sort.MOST_RELEVANT,  # urutan ulasan berdasarkan yang paling relevan
    count=20000,              # jumlah ulasan yang diambil
    filter_score_with=None    # ambil semua score 1-5
)
```

```python
import csv

with open('ulasan_mobile_jkn.csv', mode='w', newline='', encoding='utf-8') as csvfile:
  writer = csv.writer(csvfile)
  writer.writerow(['Review'])

  for review in scrapreview:
    writer.writerow([review['content']])
```

# 3. Loading Data

```python
app_reviews_df = pd.DataFrame(scrapreview)
app_reviews_df.shape
```

# 4. Menghapus data duplikat

```python
clean_df = app_reviews_df.dropna()
clean_df = clean_df.drop_duplicates()
jumlah_ulasan_setelah_hapus_duplikat, jumlah_kolom_setelah_hapus_duplikat = clean_df.shape

clean_df.info()
```
